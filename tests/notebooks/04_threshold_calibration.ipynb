```python
# Single-block robust threshold & calibration cell
# - safe: switches to project root if notebook opened from notebooks/
# - uses existing sample files (data/raw/sample_raw.csv, data/processed/sample_processed.csv)
# - writes artifacts only to repo-root/artifacts/ (not notebooks/artifacts/)
import os
from pathlib import Path
import json
import numpy as np
import pandas as pd
import joblib
import matplotlib.pyplot as plt

from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    precision_recall_curve,
    average_precision_score,
    roc_curve,
    auc,
    classification_report,
)
from sklearn.calibration import calibration_curve

# Ensure working dir is project root (safe for VSCode notebooks opened inside notebooks/)
cwd = Path.cwd()
if cwd.name == "notebooks":
    os.chdir("..")
    print("Changed working directory to project root:", Path.cwd())

# Paths (root-level)
DATA_RAW = Path("data/raw")
DATA_PROCESSED = Path("data/processed")
ARTIFACTS = Path("artifacts")
MODELS_DIR = ARTIFACTS / "models"
REPORTS_DIR = ARTIFACTS / "reports"
PLOTS_DIR = REPORTS_DIR / "plots"

# create artifact dirs at repo root (if missing) - safe and intentional
for p in (ARTIFACTS, MODELS_DIR, REPORTS_DIR, PLOTS_DIR):
    p.mkdir(parents=True, exist_ok=True)

# Choose dataset in this preference order:
# 1) data/raw/creditcard.csv (full)  2) data/raw/sample_raw.csv  3) data/processed/sample_processed.csv  4) synth fallback
full = DATA_RAW / "creditcard.csv"
sample_raw = DATA_RAW / "sample_raw.csv"
sample_processed = DATA_PROCESSED / "sample_processed.csv"

def synth_demo(n=2000, seed=42):
    rng = np.random.default_rng(seed)
    V = rng.normal(size=(n, 28))
    df = pd.DataFrame({f"V{i+1}": V[:, i] for i in range(28)})
    df["Time"] = rng.integers(0, 60*60*24*30, size=n).astype(float)
    df["Amount"] = np.expm1(rng.normal(loc=2.0, scale=1.5, size=n))
    fraud_rate = 0.03 if n < 5000 else 0.0017
    df["Class"] = rng.choice([0, 1], size=n, p=[1 - fraud_rate, fraud_rate]).astype(int)
    return df

# Load dataset (no writes to data/*)
if full.exists():
    print(f"Loading full dataset: {full}")
    df = pd.read_csv(full)
elif sample_raw.exists():
    print(f"Loading sample raw dataset: {sample_raw}")
    df = pd.read_csv(sample_raw)
    if "Class" not in df.columns:
        df["Class"] = 0
elif sample_processed.exists():
    print(f"Loading sample processed dataset: {sample_processed}")
    df = pd.read_csv(sample_processed)
    if "Class" not in df.columns:
        df["Class"] = 0
else:
    print("No dataset found in data/raw or data/processed. Creating a small synthetic demo dataset (in-memory).")
    df = synth_demo(n=2000, seed=42)

print("Dataset shape:", df.shape)
display(df.head())

# Helper functions
def evaluate_at_threshold(y_true, y_probs, thr=0.5):
    preds = (y_probs >= thr).astype(int)
    rpt = classification_report(y_true, preds, digits=4, output_dict=True)
    return {
        "threshold": float(thr),
        "precision": float(rpt.get("1", {}).get("precision", 0.0)),
        "recall": float(rpt.get("1", {}).get("recall", 0.0)),
        "f1": float(rpt.get("1", {}).get("f1-score", 0.0)),
    }

def find_best_threshold(y_true, y_probs):
    prec, rec, thresh = precision_recall_curve(y_true, y_probs)
    f1 = 2 * (prec * rec) / (prec + rec + 1e-12)
    best_idx = np.nanargmax(f1)
    best = {
        "best_threshold": float(thresh[best_idx]) if best_idx < len(thresh) else 0.5,
        "best_f1": float(f1[best_idx]),
        "precision": float(prec[best_idx]),
        "recall": float(rec[best_idx]),
    }
    return best

# Prepare X, y
if "Class" not in df.columns:
    raise ValueError("Loaded dataset has no 'Class' column.")
X = df.drop(columns=["Class"])
y = df["Class"].astype(int)

# Try load saved model(s) from top-level artifacts, else train quick RF (demo only)
model = None
for p in (MODELS_DIR / "rf_pipeline.joblib", MODELS_DIR / "xgb_model.joblib", MODELS_DIR / "logreg_pipeline.joblib"):
    if p.exists():
        try:
            model = joblib.load(p)
            print("Loaded model from:", p)
            break
        except Exception as e:
            print(f"Found {p} but failed to load: {e}")

if model is None:
    print("No pre-saved model found in artifacts/models/. Training a quick RandomForest (demo).")
    # small quick train to speed demo
    Xtrain, Xtest, ytrain, ytest = train_test_split(X, y, test_size=0.2, stratify=y if y.sum()>0 else None, random_state=42)
    model = RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1)
    model.fit(Xtrain, ytrain)
    quick_path = MODELS_DIR / "rf_quick_demo.joblib"
    joblib.dump(model, quick_path)
    print("Saved quick demo model to:", quick_path)

# Get probabilities in a robust way
if hasattr(model, "predict_proba"):
    try:
        y_proba = model.predict_proba(X)[:, 1]
    except Exception:
        y_proba = model.predict_proba(X.values)[:, 1]
elif hasattr(model, "decision_function"):
    raw = model.decision_function(X)
    y_proba = (raw - raw.min()) / (raw.max() - raw.min() + 1e-9)
else:
    y_proba = model.predict(X).astype(float)

# Baseline and tuned thresholds
baseline = evaluate_at_threshold(y, y_proba, thr=0.5)
best = find_best_threshold(y, y_proba)
print("Baseline (thr=0.5):", baseline)
print("Best threshold by F1:", best)

# Plot and save PR, ROC, Calibration (to repo-root/artifacts/reports/plots)
PLOTS_DIR.mkdir(parents=True, exist_ok=True)

# PR Curve
prec, rec, _ = precision_recall_curve(y, y_proba)
ap = average_precision_score(y, y_proba)
plt.figure()
plt.plot(rec, prec, label=f"AP={ap:.4f}")
plt.xlabel("Recall"); plt.ylabel("Precision"); plt.title("Precision-Recall Curve"); plt.legend()
pr_path = PLOTS_DIR / "precision_recall_curve.png"
plt.savefig(pr_path, bbox_inches="tight", dpi=150)
plt.show()
print("Saved PR curve to:", pr_path)

# ROC Curve
fpr, tpr, _ = roc_curve(y, y_proba)
roc_auc = auc(fpr, tpr)
plt.figure()
plt.plot(fpr, tpr, label=f"ROC AUC={roc_auc:.4f}")
plt.plot([0,1],[0,1],"--", color="gray")
plt.xlabel("False Positive Rate"); plt.ylabel("True Positive Rate"); plt.title("ROC Curve"); plt.legend()
roc_path = PLOTS_DIR / "roc_curve.png"
plt.savefig(roc_path, bbox_inches="tight", dpi=150)
plt.show()
print("Saved ROC curve to:", roc_path)

# Calibration Curve
prob_true, prob_pred = calibration_curve(y, y_proba, n_bins=10, strategy="uniform")
plt.figure()
plt.plot(prob_pred, prob_true, "o-", label="Calibration")
plt.plot([0,1],[0,1],"--", color="gray", label="Perfect")
plt.xlabel("Mean predicted prob"); plt.ylabel("Fraction of positives"); plt.title("Calibration Curve"); plt.legend()
cal_path = PLOTS_DIR / "calibration_curve.png"
plt.savefig(cal_path, bbox_inches="tight", dpi=150)
plt.show()
print("Saved calibration curve to:", cal_path)

# Save summary JSON to reports (repo-root)
REPORTS_DIR.mkdir(parents=True, exist_ok=True)
summary = {
    "dataset_shape": df.shape,
    "baseline_threshold_0.5": baseline,
    "best_threshold_by_f1": best,
    "saved_demo_model": str((MODELS_DIR / 'rf_quick_demo.joblib').resolve()) if (MODELS_DIR / 'rf_quick_demo.joblib').exists() else None,
}
(summary_path := REPORTS_DIR / "threshold_calibration_summary.json").write_text(json.dumps(summary, indent=2))
print("Saved summary to:", summary_path)

# Final: print a compact table
display(pd.DataFrame([{"metric":"baseline_f1","value":baseline["f1"]},{"metric":"best_f1","value":best["best_f1"]}]))
```
